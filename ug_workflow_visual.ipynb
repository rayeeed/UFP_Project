```python
from graphviz import Digraph

def create_ml_workflow():
    # Initialize the graph
    # 'dpi' ensures high resolution for the PNG export
    dot = Digraph(name='ML_Workflow', comment='Machine Learning Pipeline')
    dot.attr(rankdir='TB', size='8,12')  # Top-to-Bottom direction
    dot.attr('node', shape='box', style='rounded', fontname='Helvetica', fontsize='12')
    dot.attr('edge', fontname='Helvetica', fontsize='10')

    # --- Phase 1: Feature Importance ---
    dot.node('Input', 'Input Data', shape='cylinder', style='filled', fillcolor='lightgrey')
    dot.node('RF', 'Random Forest\n(Run 100 Times)')
    dot.node('AvgImp', 'Calculate Average\nFeature Importance')
    dot.node('Top10', 'Select Top 10\nFeatures', shape='parallelogram', style='filled', fillcolor='#E1F5FE')

    # --- Phase 2: Feature Combination Screening ---
    dot.node('Combos', 'Generate Feature\nCombinations')
    dot.node('XGB_Screen', 'XGBoost Optimization\n(For each combination)')
    dot.node('SelectSets', 'Evaluate Performance\n& Select Best Feature Sets', shape='parallelogram', style='filled', fillcolor='#E1F5FE')

    # --- Phase 3: Final Model Optimization (Cluster) ---
    with dot.subgraph(name='cluster_final_models') as c:
        c.attr(label='Final Model Optimization', color='grey', style='dashed', fontcolor='grey')
        c.node('LR', 'Linear Regression')
        c.node('SVR', 'SVR')
        c.node('Lasso', 'Lasso')
        c.node('Ridge', 'Ridge')
        c.node('XGB', 'XGBoost')

    dot.node('Output', 'Final Optimized\nModels', shape='cylinder', style='filled', fillcolor='lightgrey')

    # --- Define Edges (Connections) ---
    dot.edge('Input', 'RF')
    dot.edge('RF', 'AvgImp')
    dot.edge('AvgImp', 'Top10')

    dot.edge('Top10', 'Combos')
    dot.edge('Combos', 'XGB_Screen')
    dot.edge('XGB_Screen', 'SelectSets')

    # Connect selection to all final models
    dot.edge('SelectSets', 'LR')
    dot.edge('SelectSets', 'SVR')
    dot.edge('SelectSets', 'Lasso')
    dot.edge('SelectSets', 'Ridge')
    dot.edge('SelectSets', 'XGB')

    # Connect all final models to output
    dot.edge('LR', 'Output')
    dot.edge('SVR', 'Output')
    dot.edge('Lasso', 'Output')
    dot.edge('Ridge', 'Output')
    dot.edge('XGB', 'Output')

    # --- Render Output ---
    # This will create 'ml_workflow_diagram.pdf' (vector) and 'ml_workflow_diagram.png'
    output_path = dot.render('ml_workflow_diagram', view=True, format='png', cleanup=False)
    print(f"Diagram generated at: {output_path}")

if __name__ == '__main__':
    create_ml_workflow()
```
